```python
# Function to get inputs from the user
def get_inputs():
    units = input("Enter units sold (or type 'end' to stop): ")

    if units.lower() == 'end':
        return None

    price = input("Enter price per unit (or type 'end' to stop): ")

    if price.lower() == 'end':
        return None

    target = input("Enter sales target (or type 'end' to stop): ")

    if target.lower() == 'end':
        return None

    return int(units), float(price), float(target)


# Function to calculate sales performance
def sales_performance(units, price, target):

    # Calculate revenue
    revenue = units * price

    # Check that target is not zero
    if target == 0:
        print("Sales target cannot be zero.")
        return

    # Calculate target achievement
    achievement = revenue / target

    # Display results
    print(f"Revenue: ₹{revenue:,.2f}")
    print(f"Target achievement: {achievement:.0%}")


# Main loop
while True:
    try:
        # Get inputs
        data = get_inputs()

        if data is None:
            break

        # Calculate sales performance
        sales_performance(*data)

    except ValueError:
        print("Please enter valid numbers.")
```
